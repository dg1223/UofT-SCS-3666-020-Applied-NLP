```
# Build the  positive word lists from files
with open('positive-words.txt', 'rb') as f:
         positive_set = set(nltk.word_tokenize(f.read().decode('ISO-8859-1')))

# Build the  negative word lists from files
with open('negative-words.txt', 'rb') as f:
         negative_set = set(nltk.word_tokenize(f.read().decode('ISO-8859-1')))


# Now we need to remove the headers(which are not part of the word list) from the word list files.
# we can use set difference function to get rid of the header part
positive_words = positive_set.difference(negative_set)
negative_words = negative_set.difference(positive_set)
```

## Define the Parse_Sentiment Function

```
def parse_sentiment(text):
  '''
  This is a simple sentiment classifier that calculates the positivity rate of
  a text by calculating the percentage of positive words.

  It uses the predefined positive and negative word list from 'Mining and
  Summarizing Customer Reviews'. After ignoring the neutral words (neither +ve
  nor -ve) from the text, the classifier calculates the positivity rate in
  percentage. Then it standardizes the percentage score on a scale of 0 - 5, 0
  being the most negative and 5 being the most positive.

  Example:
  total word count    = 100
  positive word count = 50 (50%)
  negative word count = 25 (25%)
  neutral word count  = 25 (25%) -> ignored, not in the predefined list

  scoring standardization:
  0 % -> 0.0
  100 % -> 5.0 (100/20)

  only positive and negative words = 50 + 25 = 75
  sentiment (positivity score) = ((50/75)*100)/20 = 3.33

  ** For a more accurate sentiment analysis, we can use SentimentIntensityAnalyzer from the NLTK library.

  https://www.nltk.org/api/nltk.sentiment.vader.html#nltk.sentiment.vader.SentimentIntensityAnalyzer

  '''

  total_word_count = len(text)
  nyt_text_tokens = nltk.word_tokenize(text.lower())

  # pre-procesing
  positive_word_count = sum(1 for token in nyt_text_tokens if token in positive_words)
  negative_word_count = sum(1 for token in nyt_text_tokens if token in negative_words)

  neutral_word_count  = total_word_count - (positive_word_count + negative_word_count)

  postive_negative_words = total_word_count - neutral_word_count

  # calculate score
  sentiment = round(((positive_word_count/postive_negative_words)*100)/20)

  print(f'sentiment scores:\n0  : most negative, unfavorable or gloomy\n2-3: neutral\n5  : most positive, favorable or optimistic\n')
  return f">> NYT article sentiment = {sentiment}"
```
